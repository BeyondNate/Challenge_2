¿Qué demuestra este notebook?
1. Representa el asteroide como un grafo.
2. Implementa BFS para encontrar una ruta entre dos zonas.
3. Recorre las zonas y recolecta hasta 10 muestras.
4. Muestra las zonas visitadas y los resultados de la simulación.
Nota: la selección por identificadores pares es provisional; no constituye una clasificación mineralógica real. Para una entrega más sólida, conviene utilizar características de las muestras y aplicar criterios explícitos de selección.

```python

# ==========================================================
# CHALLENGE 1 - MINERIA DE ASTEROIDES
# PIPELINE DE RAZONAMIENTO, BUSQUEDA Y RESULTADOS
# ==========================================================

from collections import deque

# 1. PROBLEMA
print("PROBLEMA: Exploracion de un asteroide")
print("Objetivo: encontrar y recolectar muestras minerales.")
print("Limite de recoleccion: 10 muestras.")

# 2. ESPACIO DE ESTADOS
# Cada clave representa una zona y sus conexiones.

grafo = {
    "Base": ["Zona_A", "Zona_B"],
    "Zona_A": ["Base", "Zona_C", "Zona_D"],
    "Zona_B": ["Base", "Zona_E"],
    "Zona_C": ["Zona_A"],
    "Zona_D": ["Zona_A", "Zona_E"],
    "Zona_E": ["Zona_B", "Zona_D"]
}

# Muestras disponibles por zona.
# Datos creados directamente en el codigo.

muestras = {
    "Zona_A": ["M1", "M2"],
    "Zona_B": ["M3"],
    "Zona_C": ["M4", "M5"],
    "Zona_D": ["M6"],
    "Zona_E": ["M7", "M8", "M9", "M10", "M11"]
}

# 3. OPERADOR DE BUSQUEDA BFS

def buscar_ruta(inicio, objetivo):
    cola = deque([(inicio, [inicio])])
    visitados = {inicio}

    while cola:
        zona, ruta = cola.popleft()

        if zona == objetivo:
            return ruta

        for vecino in grafo[zona]:
            if vecino not in visitados:
                visitados.add(vecino)
                cola.append((vecino, ruta + [vecino]))

    return None

ruta = buscar_ruta("Base", "Zona_E")

print("\nRUTA ENCONTRADA:")
print(" -> ".join(ruta) if ruta else "No existe ruta")

# 4. EXPLORACION DEL ASTEROIDE

limite = 10
cola = deque(["Base"])
zonas_visitadas = set()
recolectadas = []

while cola and len(recolectadas) < limite:
    zona = cola.popleft()

    if zona in zonas_visitadas:
        continue

    zonas_visitadas.add(zona)
    print(f"\nExplorando: {zona}")

    for muestra in muestras.get(zona, []):
        if len(recolectadas) >= limite:
            break

        recolectadas.append(muestra)
        print(f"  Muestra recolectada: {muestra}")

    for vecino in grafo[zona]:
        if vecino not in zonas_visitadas:
            cola.append(vecino)

# 5. CLASIFICACION ILUSTRATIVA
# En este ejemplo, el criterio se basa en el identificador.
# Luego puede sustituirse por caracteristicas minerales.

candidatas = [
    m for m in recolectadas
    if int(m[1:]) % 2 == 0
]

# 6. RESULTADOS

print("\n" + "=" * 45)
print("RESULTADOS DE LA EXPLORACION")
print("=" * 45)

print("Zonas visitadas:", sorted(zonas_visitadas))
print("Muestras recolectadas:", recolectadas)
print("Total recolectado:", len(recolectadas))
print("Muestras candidatas:", candidatas)
print("Total de candidatas:", len(candidatas))

if len(recolectadas) == limite:
    print("Estado: limite de recoleccion alcanzado.")
else:
    print("Estado: exploracion terminada.")

print("Proceso de busqueda finalizado.")
```
